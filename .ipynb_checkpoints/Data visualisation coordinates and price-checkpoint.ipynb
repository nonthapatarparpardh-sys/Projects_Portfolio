```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Load the data from the provided Excel file
file_path = 'full_combined_address_mrt_supermarkets.xlsx'
df = pd.read_excel(file_path, usecols=['LATITUDE', 'LONGITUDE', 'resale_price'])

# Define the bins for the resale_price
bins = np.linspace(df['resale_price'].min(), df['resale_price'].max(), 6)
labels = [f"{int(bins[i])}-{int(bins[i+1])}" for i in range(len(bins)-1)]
df['price_bin'] = pd.cut(df['resale_price'], bins=bins, labels=labels, include_lowest=True)

# Define colors for each bin
colors = ['blue', 'green', 'yellow', 'orange', 'red']

# Create the scatter plot
plt.figure(figsize=(10, 6))
for color, label in zip(colors, labels):
    subset = df[df['price_bin'] == label]
    plt.scatter(subset['LONGITUDE'], subset['LATITUDE'], c=color, label=label, alpha=0.5)

plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.title('Scatter Plot of Properties by Price Range')
plt.legend(title='Price Range')
plt.grid(True)
plt.show()
```
