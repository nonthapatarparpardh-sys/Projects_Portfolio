```python
import pandas as pd

# Function to convert 'flat_type' values to numbers
def convert_flat_type(flat_type):
    mapping = {'1 ROOM': 1, '2 ROOM': 2, '3 ROOM': 3, '4 ROOM': 4, '5 ROOM': 5, 'EXECUTIVE': 6, 'MULTI-GENERATION': 7}
    return mapping.get(flat_type, None)  # Returns None if the flat type is not in the mapping

# Function to find the midpoint of the 'storey_range'
def convert_storey_range(storey_range):
    low, high = storey_range.split(' TO ')
    return (int(low) + int(high)) // 2

# Function to convert 'remaining_lease' to total months
def convert_remaining_lease(lease):
    parts = lease.split(' ')
    years = int(parts[0])  # The first part is always the number of years
    months = int(parts[2]) if 'month' in parts else 0  # The third part is the number of months, if it exists
    return years * 12 + months

# Load the Excel file
file_path = 'full_combined_address_mrt_and_supermarkets.xlsx'  # Replace with your file path
df = pd.read_excel(file_path)

# Applying the conversion functions
df['flat_type'] = df['flat_type'].apply(convert_flat_type)
df['storey_range'] = df['storey_range'].apply(convert_storey_range)
df['remaining_lease'] = df['remaining_lease'].apply(convert_remaining_lease)

# Save the cleaned data to a new file
new_file_path = 'final_cleaned_data.xlsx'
df.to_excel(new_file_path, index=False)

print(df.head())
```
