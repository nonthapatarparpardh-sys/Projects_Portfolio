## Splitting the Data

Once you have cleaned and preprocessed your data, and engineered any additional features, the next step in the machine learning workflow is to split the data into training and testing sets. This is a crucial step that helps to evaluate the performance of your machine learning model on unseen data, ensuring that it has not just memorized the training data (a problem known as overfitting).

<div class="alert alert-danger">
<b>Tip: Splitting Data</b> 
<p></p>    
Oftern plitting your data before cleaning and feature engineering is considered a good practice in machine learning for several important reasons:

#### 1. Data Leakage Prevention
- **Data Leakage**: Occurs when information from the test set influences the training process, leading to overly optimistic performance estimates.
- **Prevention**: Splitting data before preprocessing ensures that any knowledge gained during data cleaning and feature engineering doesn't leak from the test set to the training set.

#### 2. Real-World Scenario Simulation
- **Objective**: A model should generalize well to new, unseen data.
- **By Splitting First**: You ensure that preprocessing and feature engineering are done independently on the training and test sets, simulating how the model will encounter data in a real-world scenario.

#### 3. Proper Model Evaluation
- **Performance Metrics**: Should reflect the model's ability to generalize to new data.
- **Splitting First**: Ensures that these metrics are calculated on data that has been processed independently of the training data, providing a more accurate evaluation.

#### 4. Preventing Overfitting
- **Overfitting**: Occurs when a model performs well on the training data but poorly on new data.
- **Independent Processing**: Ensures that the model is not inadvertently tuned to specific characteristics of the test set, reducing the risk of overfitting.

#### Summary
By splitting your data before cleaning and feature engineering, you ensure that your model is evaluated fairly, has not been influenced by the test data, and is likely to perform well on new, unseen data. This practice leads to more robust and reliable machine learning models.

</div>

### Why Split the Data?

- **Training Set**: Used to train the model and learn the underlying patterns in the data.
- **Testing Set**: Used to evaluate the model's performance on unseen data, providing an estimate of how well the model is likely to perform on future, unseen data.

## Evaluating the Model

After training your machine learning models, the next critical step is to evaluate their performance on the testing set. This helps you understand how well the models are likely to perform on unseen data. For regression tasks, there are several common evaluation metrics you can use.

### 1. Mean Absolute Error (MAE)
- **Description**: The average of the absolute differences between the predicted values and the actual values.
- **Formula**: $\text{MAE} = \frac{1}{n} \sum_{i=1}^{n} \left| y_i - \hat{y}_i \right|$
- **Interpretation**: A MAE of 0 means perfect predictions. The larger the MAE, the larger the errors in prediction.

### 2. Mean Squared Error (MSE)
- **Description**: The average of the squares of the differences between the predicted values and the actual values.
- **Formula**: $\text{MSE} = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$
- **Interpretation**: Like MAE, an MSE of 0 means perfect predictions. MSE gives more weight to larger errors.

### 3. R² Score (Coefficient of Determination)
- **Description**: Represents the proportion of the variance in the dependent variable that is predictable from the independent variable(s).
- **Formula**: $ R^2 = 1 - \frac{\sum_{i=1}^{n} (y_i - \hat{y}_i)^2}{\sum_{i=1}^{n} (y_i - \bar{y})^2} $
- **Interpretation**: R² score ranges from 0 to 1. A score closer to 1 indicates better fit.

<div class="alert alert-success">
<b>Tip:</b> 
<p></p>    
Use these metrics in conjunction to get a comprehensive view of your model's performance. R² alone might be misleading if there are outliers or if the model is overfitting.
</div>

### Calculating Evaluation Metrics

You can use scikit-learn to easily calculate these metrics.

```python

mse = mean_squared_error(y_test, y_pred)

print("Mean Squared Error:", mse)
print("R² Score:", r2)
```

Replace `model` with the name of your trained model, and `X_test` and `y_test` with your testing data.

## Hyperparameter Tuning

Most machine learning algorithms have **hyperparameters**, which are settings that can be adjusted to improve the model's performance. The process of finding the optimal set of **hyperparameters** is known as hyperparameter tuning.

### What are Hyperparameters?

- **Definition**: Hyperparameters are the configuration settings used to structure machine learning models.
- **Examples**: Learning rate, tree depth, or number of clusters.
- **Importance**: The right set of hyperparameters can lead to a more accurate model, while the wrong set can lead to underfitting or overfitting.

### Methods of Hyperparameter Tuning

1. **Grid Search**
   - **Description**: Exhaustively tests a predefined set of hyperparameter values.
   - **Use Case**: When you have a small number of hyperparameters and relatively few values to test for each.
   - **Example Code**:
     ```python
     from sklearn.model_selection import GridSearchCV
     param_grid = {'n_estimators': [3, 10, 30], 'max_features': [2, 4, 6, 8]}
     grid_search = GridSearchCV(forest_reg, param_grid, cv=5)
     grid_search.fit(X_train, y_train)
     ```

2. **Random Search**
   - **Description**: Tests a random combination of hyperparameters.
   - **Use Case**: When you have a large hyperparameter space to explore.
   - **Example Code**:
     ```python
     from sklearn.model_selection import RandomizedSearchCV
     param_distribs = {'n_estimators': randint(low=1, high=200), 'max_features': randint(low=1, high=8)}
     rnd_search = RandomizedSearchCV(forest_reg, param_distributions=param_distribs, n_iter=10, cv=5)
     rnd_search.fit(X_train, y_train)
     ```

3. **Bayesian Optimization**
   - **Description**: Models the evaluation function and chooses the next hyperparameters to evaluate by balancing exploration and exploitation.
   - **Use Case**: When function evaluations are expensive and you have a limited budget.

Hyperparameter tuning is a crucial step in the machine learning workflow. It can significantly improve the performance of your model, leading to more accurate predictions. Make sure to set aside a validation set or use cross-validation to evaluate hyperparameter settings to avoid overfitting to the training data.

## Cross-Validation

After discussing hyperparameter tuning, it's essential to understand how to obtain reliable performance estimates. This is where cross-validation comes in. 

### What is Cross-Validation?

- **Definition**: Cross-validation is a resampling procedure used to evaluate machine learning models on a limited data sample.
- **Purpose**: It provides a more accurate measure of a model's performance, especially when your dataset is not large.
- **Process**: The data is split multiple times into training and test subsets. The model is trained and evaluated for each split, and the results are averaged.

### Types of Cross-Validation

1. **K-Fold Cross-Validation**
   - **Description**: The data is divided into 'k' number of subsets (or "folds"). Each fold, in turn, is used once as a validation set, while the remaining k-1 subsets form the training set.
   - **Example Code**:
     ```python
     from sklearn.model_selection import cross_val_score
     scores = cross_val_score(model, X, y, cv=5)
     ```

2. **Leave-One-Out Cross-Validation (LOOCV)**
   - **Description**: A special case of k-fold cross-validation where k is equal to the number of data points. Each data point is used once as a test set, while the rest serve as the training set.
   - **Use Case**: Useful for small datasets.

3. **Stratified K-Fold Cross-Validation**
   - **Description**: Similar to k-fold cross-validation, but ensures that each fold has the same proportion of observations with a given target value, especially important for imbalanced datasets.

<div class="alert alert-success">
<b>Tip:</b> 
<p></p>    
While cross-validation provides a better performance estimate, it's computationally more expensive, especially for large datasets or complex models. Always balance the benefits against computational cost.
</div>

<div class="alert alert-info">
<b>Example:</b> 
<p></p>    
If you have a dataset of 100 samples and use 10-fold cross-validation, each fold will contain 10 samples. For each iteration, 90 samples will be used for training and 10 for testing.
</div>

Cross-validation is a robust method to assess the performance of machine learning models. It provides a more accurate measure by reducing the variance that can occur from a random train-test split. Especially when tuning hyperparameters, it's crucial to use cross-validation to ensure that the chosen parameters generalize well to new, unseen data.
